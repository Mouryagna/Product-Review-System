```python
import pandas as pd
import re
```

```python
df = pd.read_csv("../data/amazon_reviews_20000.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
for i, review in enumerate(df["review_text"].head(5), start=1):
    print(f"\nReview {i}:")
    print(review)
```

- Convert text to lowercase.
- Remove HTML tags.
- Remove URLs.
- Remove unnecessary punctuation while retaining letters and numbers.
- Normalize repeated whitespace.

```python
def preprocess_text(text):
    text = str(text).lower()

    text = re.sub(r"<[^>]+>", " ", text)
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    text = re.sub(r"[^a-z0-9\s']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
examples = [
    "THIS PRODUCT IS AMAZING!!!",
    "The product is NOT good.",
    "Works perfectly :) Visit https://example.com"
]

for example in examples:
    print("Original :", example)
    print("Processed:", preprocess_text(example))
    print("-" * 50)
```

```python
df["cleaned_text"] = df["review_text"].apply(preprocess_text)

df[["review_text", "cleaned_text", "sentiment"]].head(10)
```

```python
empty_reviews = df["cleaned_text"].str.strip().eq("").sum()

print("Empty reviews after preprocessing:", empty_reviews)
```

```python
print(
    "Duplicate cleaned reviews:",
    df["cleaned_text"].duplicated().sum()
)
```

```python
path="../data/amazon_reviews_preprocessed.csv"
df.to_csv(
    path,
    index=False,
    encoding="utf-8"
)

print("Preprocessed dataset saved.")
print("Shape:", df.shape)
```
